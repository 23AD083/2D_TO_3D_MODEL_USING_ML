# Data Exploration

In this notebook, we will explore the dataset used for predicting room details. We will visualize the data distributions and understand the features that will be used in our machine learning model.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set the style for seaborn
sns.set(style='whitegrid')
```

```python
# Load the dataset
data_path = '../data/processed/room_details.csv'
df = pd.read_csv(data_path)

# Display the first few rows of the dataset
df.head()
```

```python
# Summary statistics of the dataset
df.describe()
```

```python
# Visualizing the distribution of the target variable: number of days to be worked
plt.figure(figsize=(10, 6))
sns.histplot(df['days_to_work'], bins=30, kde=True)
plt.title('Distribution of Days to Work')
plt.xlabel('Days to Work')
plt.ylabel('Frequency')
plt.show()
```

```python
# Visualizing the relationship between area and days to work
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='area', y='days_to_work', hue='walls', palette='viridis')
plt.title('Area vs Days to Work')
plt.xlabel('Area')
plt.ylabel('Days to Work')
plt.legend(title='Walls')
plt.show()
```

```python
# Correlation heatmap
plt.figure(figsize=(10, 8))
correlation = df.corr()
sns.heatmap(correlation, annot=True, fmt='.2f', cmap='coolwarm', square=True)
plt.title('Correlation Heatmap')
plt.show()
```
